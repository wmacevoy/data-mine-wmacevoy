# Cereal Scatter Plot (R Jupyter Notebook)
This notebook loads `cerials.csv` and plots a scatter plot using ggplot2.

```r
library(ggplot2)
library(readr)
library(dplyr)

# Load the data
data <- read_csv('data/cerials.csv')
glimpse(data)
head(data)
```

```r
# Scatter plot example: change x and y for your data columns
ggplot(data, aes(x = Calories, y = Sugars)) +
  geom_point(color = 'steelblue', size = 3, alpha = 0.7) +
  theme_minimal(base_size = 14) +
  labs(title = 'Cereal Nutritional Scatter Plot',
       x = 'Calories per Serving',
       y = 'Sugars (g per Serving)')
```

```r
# Add a linear regression fit with 95% CI to the scatter plot.
# If your column names differ, edit x= and y= accordingly.
ggplot(data, aes(x = Calories, y = Sugars)) +
  geom_point(size = 3, alpha = 0.7) +
  geom_smooth(method = 'lm', se = TRUE) +  # 95% confidence band around the regression line
  theme_minimal(base_size = 14) +
  labs(title = 'Cereal Nutritional Scatter Plot (with Linear Fit)',
       x = 'Calories per Serving',
       y = 'Sugars (g per Serving)')
```
