# Student Guide: Evaluation

Goal: measure whether retrieval and answers are good.

What to track (starter list):
- Context recall@k: does the gold passage appear in the top‑k?
- Precision@k: how many of the retrieved are actually relevant?
- MRR / nDCG: ranking‑sensitive quality metrics.
- Answer faithfulness & relevance (can use LLM‑as‑judge for small studies).
- Latency: retrieval + generation time.

References:
- BEIR benchmark: https://github.com/beir-cellar/beir
- DCG/nDCG: https://en.wikipedia.org/wiki/Discounted_cumulative_gain

# 05 – Evaluation

Sketch of basic metrics to track for retrieval quality and latency.


```python
# Example placeholders for evaluation metrics
from dataclasses import dataclass
from typing import List

@dataclass
class RetrievalResult:
    relevant: bool

# Fake ground truth and results for illustration
truth: List[bool] = [True, False, True, True, False]
results: List[RetrievalResult] = [RetrievalResult(b) for b in truth]

context_recall = sum(r.relevant for r in results) / len(results)
print({"context_recall": round(context_recall, 3)})
```
