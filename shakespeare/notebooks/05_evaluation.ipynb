# 05 – Evaluation

Sketch of basic metrics to track for retrieval quality and latency.


```python
# Example placeholders for evaluation metrics
from dataclasses import dataclass
from typing import List

@dataclass
class RetrievalResult:
    relevant: bool

# Fake ground truth and results for illustration
truth: List[bool] = [True, False, True, True, False]
results: List[RetrievalResult] = [RetrievalResult(b) for b in truth]

context_recall = sum(r.relevant for r in results) / len(results)
print({"context_recall": round(context_recall, 3)})
```
