# Student Guide: Embeddings

Goal: turn cleaned text chunks into vectors we can search.

What you’ll learn:
- How TF–IDF builds a sparse bag‑of‑words matrix.
- How TruncatedSVD (LSA) produces dense semantic vectors.
- Why we row‑normalize vectors for cosine similarity.

Algorithm overview:
1) Fit `TfidfVectorizer` on the corpus (vocabulary + sparse matrix).
2) Fit `TruncatedSVD` on TF–IDF to get `dense = normalize(SVD(tfidf))`.
3) Persist artifacts so we can query later without re‑fitting.

References:
- TF–IDF: https://scikit-learn.org/stable/modules/feature_extraction.html#tfidf-term-weighting
- LSA / TruncatedSVD: https://scikit-learn.org/stable/modules/decomposition.html#lsa
- Cosine similarity: https://scikit-learn.org/stable/modules/metrics.html#cosine-similarity
- Modern alternatives (transformer embeddings): https://www.sbert.net/
